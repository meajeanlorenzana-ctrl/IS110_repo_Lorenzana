A. Create an ndarry object (A) with any values from 1 to 100 in a nested Python List. You can choose your range from the given values (1-100).

```python
import numpy as np

#the import as numpy Loads the Numerical Python library, enabling high-performance multidimensional array objects and advanced mathematical tools.
#whille the as np is assign as alias or nickname
```

```python
nested_list = [[10, 20], [40, 50], [80, 90]]

# this is the raw data structure before it is converted into numpy array
```

```python
A = np.array(nested_list)

#this is the resulting ndarray object optimized for numerical processing
```

1. Using a property of the ndarray object
      - e.g. `A.shape`

```python
print(A.shape)

#This returns the (3, 2) arrays, representing the dimensions (rows, columns)
```

2. Using a method of the ndarray object
      - e.g. `A.max()`

```python
print(A.max())

#this is the funtion that finds the maximum value
```

3. Using a numpy function and passing an ndarray
     - e.g. `np.max(A)`

```python
print(np.max(A))

#this is the funtion that used to find the maximum value
```

B. Create a 5x5 integer-array filled with ones.

```python
B = np.ones((5, 5), dtype=int)

#Initializes a 5x5 grid of integer 1s
```

```python
print(B.shape)

#Returns the shape (5, 5) - Beside: Checking dimensions
```

```python
print(B.size)

#Returns the total element count (25) - Beside: Checking size
```

```python
print(B.ndim)

#Returns the number of dimensions (2) - Beside: Checking axes
```
