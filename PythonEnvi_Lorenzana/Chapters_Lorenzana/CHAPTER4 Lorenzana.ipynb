#### Exercise 1

Rewrite your pay computation with time-and-a-half for overtime and create a function called compute pay which takes two parameters (hours and rate).

    Enter Hours: 45
    Enter Rate: 10
    Pay: 475.0
    
[PEW] E4.6

```python
def computepay(hours, rate):
    if hours > 40:
        reg_pay = 40 * rate
        ot_pay = (hours - 40) * (rate * 1.5)
        return reg_pay + ot_pay
    else:
        return hours * rate

try:
    h = float(input("Enter Hours: "))
    r = float(input("Enter Rate: "))
    print("Pay:", computepay(h, r))
except ValueError:
    print("Error: Please enter numeric values.")
```

## CHALLENGE

Write a function called computegrade that takes a score as its parameter and returns a grade as a string using the following table:

    Score           Grade
    [ 98, 100]      1.0
    [ 94,  97)		1.25
    [ 90,  93)		1.5
    [ 86,  89)		1.75
    [ 82,  85)		2.0
    [ 78,  81)		2.25
    [ 75,  77)		2.5
    [ 0,   74)		5.0

    Run the program repeatedly to test the various different values for input.
    
<br>

```python
def computegrade(score):
    if score >= 98 and score <= 100:
        return "1.0"
    elif score >= 94:
        return "1.25"
    elif score >= 90:
        return "1.5"
    elif score >= 86:
        return "1.75"
    elif score >= 82:
        return "2.0"
    elif score >= 78:
        return "2.25"
    elif score >= 75:
        return "2.5"
    elif score >= 0:
        return "3.0"
    else:
        return "Invalid Score"

while True:
    user_input = input("Enter score ('quit'): ")
    if user_input.lower() == 'quit':
        break
    try:
        score_val = float(user_input)
        print("Grade:", computegrade(score_val))
    except ValueError:
        print("Error: Please enter a number.")
```
