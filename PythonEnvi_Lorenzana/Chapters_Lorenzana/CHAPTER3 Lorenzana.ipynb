#### Exercise 2

Write your salary computation to give an employee 1.5 times the hourly rate for hours worked above 40 hours. Example:

    Enter Hours: 45
    Enter Rate: 10
    Pay: 475.0


````python
hours = input("Enter Hours: ")
rate = input("Enter Rate: ")

h = float(hours)
r = float(rate)

if h > 40:
    reg_pay = 40 * r

    overtime_h = h - 40
    ot_pay = overtime_h * (r * 1.5)
    pay = reg_pay + ot_pay
else:
    pay = h * r

    print("Pay:", pay)
````

## CHALLENGE

Write a program to prompt for a score between 0 and 100 and print a grade using the following table:

    Score           Grade
    [ 98, 100]      1.0
    [ 94,  97)		1.25
    [ 90,  93)		1.5
    [ 86,  89)		1.75
    [ 82,  85)		2.0
    [ 78,  81)		2.25
    [ 75,  77)		2.5
    [ 0,   74)		5.0

Adapted from [PEW] E3.3
<br>

````python
try:
    score = float(input("Enter Score: "))

    if score >= 98 and score <= 100:
        print("1.0")
    elif score >= 94:
        print("1.25")
    elif score >= 90:
        print("1.5")
    elif score >= 86:
        print("1.75")
    elif score >= 82:
        print("2.0")
    elif score >= 78:
        print("2.25")
    elif score >= 75:
        print("2.5")
    elif score >= 0:
        print("3.0")
    else:
        print("Score out of range")
except:
    print("Please enter a number")
````
